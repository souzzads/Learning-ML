# Métodos Descritivos e Preditivos

## 1. Métodos Descritivos

Os métodos descritivos são utilizados para **entender e resumir dados que já foram coletados**.

Eles respondem principalmente:

> **"O que aconteceu?"**

O objetivo é encontrar padrões, comportamentos, tendências e informações relevantes nos dados.

### Exemplos

- Média, mediana e moda
- Mínimo e máximo
- Porcentagens
- Distribuições
- Gráficos
- Agrupamentos e comparações
- Análise de vendas por período, produto ou região

### Exemplo

Uma empresa possui dados de vendas de 2025.

Podemos descobrir:

- Qual produto vendeu mais?
- Qual cidade teve maior faturamento?
- Qual foi o mês com maior número de vendas?
- Qual foi o lucro médio?

Nesse caso, estamos **descrevendo o que já aconteceu**.

---

## 2. Métodos Preditivos

Os métodos preditivos utilizam **dados históricos para tentar prever resultados futuros ou desconhecidos**.

Eles respondem principalmente:

> **"O que provavelmente vai acontecer?"**

Normalmente utilizam técnicas de **Machine Learning, estatística e modelos matemáticos**.

### Exemplos

- Regressão Linear
- Regressão Logística
- Árvores de Decisão
- Random Forest
- Redes Neurais
- Modelos de séries temporais

### Exemplo

Uma empresa possui dados de vendas dos últimos 3 anos.

Com esses dados, podemos criar um modelo para prever:

- Quanto será vendido no próximo mês?
- Qual cliente tem maior chance de cancelar?
- Qual produto terá maior demanda?
- Qual será o faturamento futuro?

Nesse caso, estamos tentando **prever algo que ainda não aconteceu**.

---

## 3. Diferença principal

| Método | Pergunta principal | Objetivo |
|---|---|---|
| Descritivo | O que aconteceu? | Entender os dados |
| Preditivo | O que pode acontecer? | Fazer previsões |

### Exemplo simples

Imagine uma loja:

**Descritivo:**
> "Em setembro, a loja vendeu R$ 100.000."

**Preditivo:**
> "Com base nos dados anteriores, o modelo prevê vendas de aproximadamente R$ 110.000 em outubro."

---

## 4. Resumindo

**Métodos descritivos → entender o passado e o presente.**

**Métodos preditivos → usar dados para estimar o futuro ou resultados desconhecidos.**

Na Ciência de Dados, os dois são importantes: primeiro podemos **entender os dados**, depois podemos utilizar essas informações para **criar modelos preditivos**.

# Classificação

A **classificação** é um tipo de problema de **Machine Learning supervisionado**.

Seu objetivo é fazer o modelo aprender a **classificar dados em categorias ou classes**.

> **Pergunta principal:** "A qual categoria esse dado pertence?"

---

## 1. Como funciona?

No aprendizado supervisionado, o modelo recebe dados que já possuem uma **resposta conhecida (rótulo)**.

### Exemplo

Uma empresa possui dados de transações:

| Valor | Localização | Horário | Fraude |
|---:|---|---|---|
| R$ 50 | Sorocaba | 14:00 | Não |
| R$ 5.000 | São Paulo | 03:00 | Sim |
| R$ 80 | Votorantim | 15:00 | Não |

O modelo aprende com esses exemplos.

Depois, recebe uma nova transação:

> R$ 4.000 → São Paulo → 02:30

E tenta classificar:

> **Fraude: Sim**

---

## 2. Classes

As categorias possíveis são chamadas de **classes**.

### Classificação binária

Possui apenas **duas classes**.

Exemplos:

- Fraude / Não fraude
- Aprovado / Reprovado
- Doente / Saudável
- Spam / Não spam

### Classificação multiclasse

Possui **mais de duas classes**.

Exemplo:

Uma IA pode classificar uma imagem como:

- Cachorro
- Gato
- Cavalo
- Pássaro

---

## 3. Algoritmos de classificação

Alguns algoritmos utilizados para classificação:

- Regressão Logística
- K-Nearest Neighbors (KNN)
- Árvore de Decisão
- Random Forest
- Support Vector Machine (SVM)
- Redes Neurais

---

## 4. Classificação x Regressão

Os dois são problemas de **aprendizado supervisionado**, mas possuem objetivos diferentes.

| Tipo | Resultado |
|---|---|
| **Classificação** | Categoria |
| **Regressão** | Valor numérico |

### Exemplo

**Classificação:**

> Essa transação é fraude ou não?

**Regressão:**

> Qual será o valor da próxima venda?

---

## 5. Resumindo

**Classificação = prever uma categoria.**

O modelo recebe exemplos com respostas conhecidas, aprende os padrões e utiliza esse conhecimento para classificar novos dados.

> **Dados + rótulos → treinamento → modelo → nova entrada → classe prevista**
